**Filter Based on String**

```python
data = {"Name":["Alice","Bob","Charlie","David","Eve","Frank","Grace","Hannah","Raya"],
        "City":["New York","Los Angeles","Newark","Boston","New Delhi","London","New Orleans","Houston","New London"],
        "Department":["HR","IT","Finance","IT","HR","Marketing","Finance","HR","Marketing"],
        "Salary":[50000,60000,55000,70000,52000,58000,62000,51000,65000]}

df = pd.DataFrame(data)
df
```

```python
# Searching for rows where city name is "London"

df[df.loc[:,"City"].str.contains("London")]
```

```python
df.loc[df["City"].str.contains("London")]
```

```python
# case insensitive search
df.loc[df["City"].str.contains("london",case=False)]
```

```python
# searching for rows where city name starts with "new"
df[df["City"].str.contains(r"^new",case=False)]
```

```python
# Searching for rows where city name ends with 'rk'
df[df["City"].str.contains(r"rk$")]
```

```python
# Searching for rows where person's name starts with a vowel
df[df["Name"].str.contains(r"^[aeiou]",case=False)]
```

```python
# Using conditions inside regular expression
df[df["City"].str.contains(r"new|los",case=False)]
```

**Adding New Column**

```python
df = pd.read_csv('student_data.csv')
df.head()
```

```python
# Adding a column with a constant value
df["Country"] = "Bangladesh"
df.head()
```

```python
# Adding arithmetically modified column
df["Total Marks"] = df ["Data Structure Marks"] + df["Algorithm Marks"] + df["Python Marks"]
df.head()
```

```python
# Adding conditionally modified column
df["Grades"] = np.where(df["Total Marks"]>=240,"A+","A")
df.head()
```

```python
# Adding column via string operation
df["First Name"] = df["FullName"].str.split(" ").str[0]
df["Last Name"] = df['FullName'].str.split(" ").str[1]
df.head()
```

**Datetime Functions of Pandas**

```python
df = pd.read_csv("student_completed_data.csv")
df
```

```python
df["EnrollmentDate"]
```

```python
df["EnrollmentDate"] = pd.to_datetime(df["EnrollmentDate"])
df["FinishedDate"] = pd.to_datetime(df["FinishedDate"])
```

```python
df["Time Duration"] = df["FinishedDate"] - df["EnrollmentDate"]
df["Month"] = df["EnrollmentDate"].dt.month_name()
df["Day name"] = df["EnrollmentDate"].dt.day_name()
df.sort_values(by=["Time Duration"])
```
