```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic_dataset.csv")
df.info()
```

### **Categorical Value Analysis**

```python
sns.countplot(data=df, x="Survived")
```

```python
survival_rate = (df["Survived"].value_counts() / df["Survived"].value_counts().sum()) * 100
survival_rate
```

```python
plt.pie(survival_rate, labels=survival_rate.index, autopct="%0.2f%%")
```
