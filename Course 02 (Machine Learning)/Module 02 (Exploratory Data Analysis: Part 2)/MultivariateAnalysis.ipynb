```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic_dataset.csv")
```

### **Categorical Analysis**

```python
sns.countplot(x=df["Sex"], hue=df["Survived"])
```

```python
gender_split = df.groupby("Sex")["Survived"].value_counts(normalize=True)
gender_split
```

```python
sns.countplot(x=df["Pclass"], hue=df["Survived"])
```

```python
class_split = df.groupby("Pclass")["Survived"].value_counts(normalize=True)
class_split
```

```python
sns.displot(x=df["Survived"], y=df["Embarked"])
```

### **Categorical Analysis with Numerical Values**

```python
sns.barplot(x=df["Pclass"], y=df["Fare"], hue=df["Sex"])
```

```python
class_fare = df.groupby("Pclass")["Fare"].mean()
class_fare
```

```python
sns.barplot(x=df["Survived"], y=df["Age"], hue=df["Sex"])
```

```python
survival_age = df.groupby(["Survived","Sex"])["Age"].mean()
survival_age
```

### **KDE Plot for Categorical Discrepancy**

```python
sns.kdeplot(x=df["Age"], hue=df["Survived"])
```

```python
sns.kdeplot(x=df["Age"], hue=df["Sex"])
```
