```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("marks_dataset.csv")
```

```python
sns.kdeplot(df["maths_marks"])
```

```python
sns.boxplot(df["maths_marks"])
```

```python
df["maths_marks"].describe()
```

```python
x = 12
x = x/100

min_range = df["maths_marks"].quantile(x)
max_range = df["maths_marks"].quantile(1-x)

print(f"min range: {min_range}, max range: {max_range}")
```

```python
df["maths_marks"] = df["maths_marks"].clip(min_range, max_range)
df["maths_marks"].describe()
```
