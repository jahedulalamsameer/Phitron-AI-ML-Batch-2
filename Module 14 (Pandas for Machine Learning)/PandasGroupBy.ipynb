```python
import pandas as pd
import numpy as np
```

**Pandas GroupBy**

```python
df = pd.read_csv("student_data.csv")
df["Total"] = df.iloc[:,2:5].sum(axis=1)
df.drop(columns=["Data Structure Marks","Algorithm Marks","Python Marks","EnrollmentDate"],inplace=True)
df.loc[df["Total"]<=200,"Total"] = np.nan
df["CompletionStatus"] = "Completed"
df
```

```python
instructor_group = df.groupby("Instructor")
instructor_group
```

```python
instructor_group.describe()
```

```python
instructor_group["Total"].mean()
```

```python
df["Total"] = df["Total"].fillna(df["Instructor"].map(instructor_group["Total"].mean()))
df
```
