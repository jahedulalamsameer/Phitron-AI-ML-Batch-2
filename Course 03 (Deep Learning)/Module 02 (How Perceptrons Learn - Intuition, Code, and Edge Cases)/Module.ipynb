## **Perceptrons from scratch**

### **Main Implementation**

```python
class Perceptron:
    def __init__(self, eta, epochs):
        self.weights = np.random.randn(3) * 1e-4
        print(f"Weights: \n{self.weights}\n")

        self.eta = eta
        self.epochs = epochs

    def activation_func(self, inputs, weights):
        z = np.dot(inputs, weights)
        return np.where(z<1, 0, 1)

    def fit(self, X, y):
        self.X = X
        self.y = y

        X_with_bias = np.c_[self.X, -np.ones((len(self.X),1))]
        print(f"X with bias: \n{X_with_bias}\n")

        for epoch in range(self.epochs):
            print((f"Epoch: {epoch}"))

            y_hat = self.activation_func(X_with_bias, self.weights)
            print((f"Predictions: \n{y_hat}"))

            error = self.y - y_hat
            print(f"Error: \n{error}")

            self.weights = self.weights + (self.eta * np.dot(X_with_bias.T, error))
            print(f"Updated weights: \n{self.weights}")

            print("#############\n")


    def predict(self, X):
        X_with_bias = np.c_[self.X, -np.ones((len(self.X),1))]

        return self.activation_func(X_with_bias, self.weights)
```

---

### **AND Gate**

```python
AND = {
    "x1" : [0,0,1,1],
    "x2" : [0,1,0,1],
    "y" : [0,0,0,1]
}

AND = pd.DataFrame(AND)
AND
```

```python
X = AND.drop("y", axis=1)
y = AND["y"]
y.to_frame()
```

```python
and_model = Perceptron(eta=0.5, epochs=10)
and_model.fit(X,y)
```

### **OR Gate**

```python
OR = {
    "x1" : [0,0,1,1],
    "x2" : [0,1,0,1],
    "y" : [0,1,1,1]
}

OR = pd.DataFrame(OR)

X = OR.drop("y", axis=1)
y = OR["y"]
y.to_frame()

or_model = Perceptron(eta=0.5, epochs=10)
or_model.fit(X,y)
```

### **XOR Gate**

```python
XOR = {
    "x1" : [0,0,1,1],
    "x2" : [0,1,0,1],
    "y" : [0,1,1,0]
}

XOR = pd.DataFrame(XOR)

X = XOR.drop("y", axis=1)
y = XOR["y"]
y.to_frame()

xor_model = Perceptron(eta=0.1, epochs=1000)
xor_model.fit(X,y)
```
