# **PyTorch NN Module**
---

## **Importing Necessary Libraries, Modules and Packages**

```python
import torch
import torch.nn as nn
```

```python
# Creating dataset
features = torch.rand(10,5)
features
```

---

## **Version 1**

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out = self.linear(features)
        out = self.sigmoid(out)

        return out
```

```python
# Creating model
model = Model(features.shape[1])

# Forward Pass
model.forward(features)
```

```python
# Showing model weights
model.linear.weight
```

```python
# Showing model bias
model.linear.bias
```

```python
# Visualize
!pip install torchinfo
```

```python
from torchinfo import summary
summary(model, input_size=(10,5))
```

## **Version 2**

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        # Layer 1
        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()

        # Layer 2
        self.linear2 = nn.Linear(3,1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        # Layer 1
        out = self.linear1(features)
        out = self.relu(out)

        # Layer 2
        out = self.linear2(out)
        out = self.sigmoid(out)

        return out
```

```python
# Creating model
model = Model(features.shape[1])

# Forward Pass
model.forward(features)
```

```python
# Showing model weights
model.linear1.weight
```

```python
# Showing model bias
model.linear1.bias
```

```python
summary(model, input_size=(10,5))
```

---

## **Version 3**

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(num_features, 3), # index 0
            nn.ReLU(), # index 1
            nn.Linear(3,1), # index 2
            nn.Sigmoid() # index 3
        )

    def forward(self, features):
        out = self.network(features)
        return out
```

```python
# Creating model
model = Model(features.shape[1])

# Forward Pass
model.forward(features)
```

```python
# Showing model weights
model.network[0].weight
```

```python
model.network[2].weight
```

```python
# Showing model bias
model.network[0].bias
```

```python
model.network[2].bias
```

```python
# Visuualize
summary(model, input_size=(10,5))
```
