```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

```python
df = pd.read_csv("staff_data.csv")

arr_emp_ages = np.array(df["emp_age"])
arr_emp_salary = np.array(df["emp_salary"])

print(f"Array of emp_ages:\n{arr_emp_ages}")
print(f"Datatype: {arr_emp_ages.dtype}")
print(f"Number of dimesions: {arr_emp_ages.ndim}")
print(f"Shape: {arr_emp_ages.shape}")
print(f"Total number of elements: {arr_emp_ages.size}\n")

print(f"Array of emp_ages:\n{arr_emp_salary}")
print(f"Datatype: {arr_emp_salary.dtype}")
print(f"Number of dimesions: {arr_emp_salary.ndim}")
print(f"Shape: {arr_emp_salary.shape}")
print(f"Total number of elements: {arr_emp_salary.size}")
```

**Q2**

```python
max_salary = arr_emp_salary.max()
min_salary = arr_emp_salary.min()
avg_salary = arr_emp_salary.mean()
avg_age = arr_emp_ages.mean()

print(f"Maximum salary: {max_salary}")
print(f"Minimum salary: {min_salary}")
print(f"Average salary: {avg_salary}")
print(f"Average age: {avg_age}")
```

```python
index_ages = (arr_emp_ages >= 25) & (arr_emp_ages <= 40)

ages_between_25_40 = arr_emp_ages[index_ages == True]
print(f"Employees aged between 25 & 40:\n{ages_between_25_40}\n")

index_salary = arr_emp_salary > 50000
salary_above_50k = arr_emp_salary[index_salary == True]
print(f"Employees above 50,000 salary:\n{salary_above_50k}\n")

print(f"The number of fulfiling both conditions: {ages_between_25_40.size + salary_above_50k.size}")
```

```python
updated_arr_emp_ages = arr_emp_ages.copy()

for i in range(len(updated_arr_emp_ages)):
  if updated_arr_emp_ages[i]<avg_age:
    updated_arr_emp_ages[i]+=4
  elif updated_arr_emp_ages[i]>=avg_age:
    updated_arr_emp_ages[i]+=2


updated_avg_age = updated_arr_emp_ages.mean()

print(f"Original array:\n{arr_emp_ages}\n")
print(f"Updated array:\n{updated_arr_emp_ages}\n")
print(f"Updated average age: {updated_avg_age}")
```
