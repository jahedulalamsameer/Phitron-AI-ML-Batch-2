```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Practice.csv")
df
```

**Problem 1**

```python
df["DiscountedPrice"] = df["Price"] - (df["Price"]*0.10)
df
```

**Problem 2**

```python
df["EmployeeID"].loc[df.duplicated()==True].count()
```

```python
df.drop_duplicates(inplace=True)
df
```

**Problem 3**

```python
df.info()
```

```python
# Filling missing age with avg age

df.fillna({"Age":int(df["Age"].mean())},inplace=True)
df
```

```python
# Dropping rows with missing salary

df.dropna(subset=["Salary"],inplace=True)
df
```

**Problem 4**

```python
print(f"""Marks1 mean: {df["Marks1"].mean():.2f}
Marks2 mean: {df["Marks2"].mean():.2f}
Marks3 mean: {df["Marks3"].mean():.2f}""")
```

```python
df["TotalMarks"] = df.iloc[::,5:8].sum(axis=1)
df
```

```python
print(f"Standard Deviation: {df["TotalMarks"].std():.4f}")
```
