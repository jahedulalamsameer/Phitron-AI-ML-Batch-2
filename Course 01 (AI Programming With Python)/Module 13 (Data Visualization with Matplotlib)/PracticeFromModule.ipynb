```python
# Visualizing pandas dataframe

df = pd.read_csv("enrollment_data.csv")
df
```

**Bar Chart**

```python
df = pd.read_csv("student_data.csv")
df.head()
```

```python
# By instructor

instructor = df.groupby("Instructor").size()
instructor.index
```

```python
plt.title("Number of Students per Instructors")
plt.xlabel("Instructors")
plt.ylabel("Number of Students")

plt.bar(instructor.index,instructor.values,color=["#11b86f","#5f1207ec","#020e25ff"])
plt.show()
```

```python
# By completion status

status = df.groupby(by="CompletionStatus").size()

plt.title("Number of Students per Completion Status")
plt.xlabel("Completion Status")
plt.ylabel("Number of Students")

plt.bar(status.index,status.values,color=["#020e25ff","#11b86f","#5f1207ec"])
plt.show()
```

**Pie Chart**

```python
# By completion status

plt.pie(status,labels=status.index,autopct="%0.1f%%",explode=(0.01,0.02,0.02))
plt.show()
```
