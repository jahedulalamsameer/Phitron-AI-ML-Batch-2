```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("sns_data.csv")
df.columns
```

### **Line Plot using Seaborn**

Axes Level Plotting

```python
sns.lineplot(data=df,x="week",y="attendance_rate",hue="gender",errorbar=None)
```

```python
sns.lineplot(data=df,x="week",y="attendance_rate",hue="subject",errorbar=None)
```

Figure Level Plotting

```python
sns.relplot(kind="line",data=df,x="week",y="attendance_rate",hue="subject",errorbar=None)
```

### **Scatterplot using Seaborn**

```python
student = pd.read_csv("students.csv")
student.head()
```

Axes Level

```python
sns.scatterplot(data=student, x="study_hours", y="test_score", hue="gender", style="subject", size="class_level")
```

Figure Level

```python
sns.relplot(kind="scatter", data=student, x="study_hours", y="test_score", hue="gender", style="subject", size="class_level")
```

```python
tips_data = sns.load_dataset("tips")
tips_data
```

```python
sns.relplot(kind="scatter",data=tips_data,x="total_bill",y="tip",hue="sex",size="size",style="time")
```

### **Facet Plot using Seaborn**

```python
student.columns
```

```python
sns.relplot(kind="scatter", data=student, x="study_hours", y="test_score", hue="gender")
```

```python
sns.relplot(kind="scatter", data=student, x="study_hours", y="test_score", col="gender")
```

```python
sns.relplot(kind="scatter", data=student, x="study_hours", y="test_score", col="gender", row="hostel")
```

```python
sns.relplot(kind="scatter", data=student, x="study_hours", y="test_score", col="week",col_wrap=3)
```

### **Histogram using Seaborn**

Axes Level Plotting

```python
student.columns
```

```python
sns.histplot(data=student, x="attendance_rate", hue="gender", bins=10)
```

Figure Level Plotting

```python
sns.displot(kind="hist", data=student, x="attendance_rate", hue="gender", col="gender")
```

```python
sns.displot(kind="hist", data=student, x="attendance_rate", hue="gender", col="gender", row="hostel")
```

### **KDE Plot using Seaborn**

```python
sns.histplot(data=tips_data, x="tip")
```

```python
sns.kdeplot(data=tips_data, x="tip", hue="sex", fill=True)
```

### **Countplot using Seaborn**

```python
student.columns
```

```python
sns.countplot(data=student, x="subject", hue="gender")
```

### **Bar Plot using Seaborn**

```python
student.columns
```

```python
sns.barplot(data=student, x="gender", y="Marks", hue="subject", errorbar=None)
```

```python
sns.barplot(data=student, x="gender", y="Marks", hue="subject", errorbar=None, estimator=np.median)
```

### **Regression Plot using Seaborn**

Axes Level Plotting

```python
sns.regplot(data=student, x="study_hours", y="test_score")
```

Figure Level Plotting

```python
sns.lmplot(data=student, x="study_hours", y="test_score", hue="hostel")
```

```python
sns.lmplot(data=student, x="study_hours", y="test_score", row="hostel", col="gender")
```
