```python
with open('sample.txt','r') as f:
    strings = f.readlines()
    lines = len(strings)
    

    words=0; characters=0
    for line in strings:
        line = line.strip().split()
        words += len(line)
        for word in line:
            characters += len(word)
    
    
with open('counter_of_string.txt','w') as f:
    f.writelines(f"Number of lines: {lines}\nNumber of words: {words}\nNumber of characters: {characters}")
print(lines)
print(words)
print(characters)
```

```python
with open('names.txt','r') as f1:
    lst = f1.readlines()
    
    lst = list(map(lambda x: x.strip().lower().title(), lst))
    lst = list(set(lst))
    lst.sort()
    

with open('clean_names.txt','w') as f2:
    f2.writelines(list(map(lambda x: x+"\n",lst)))
```
